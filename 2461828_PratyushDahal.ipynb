# 3. Outlier Detection:

```python
#Detecting outliers in HDI and GNI per Capita using the 1.5 × IQR rule.
#at first we need to find q1 n q3 of hdi and then subtracting them we get iqr

q1_of_HDI = south_asia_df['hdi'].quantile(0.25)
q3_of_HDI = south_asia_df['hdi'].quantile(0.75)

#IQR OF HDI
IQR_HDI = q3_of_HDI - q1_of_HDI
print(IQR_HDI)


#at seconf we need to find q1 n q3 of GNI and then subtracting them we get iqr

q1_of_GNI = south_asia_df['gross_inc_percap'].quantile(0.25)
q3_of_GNI = south_asia_df['gross_inc_percap'].quantile(0.75)

#IQR OF HDI
IQR_GNI = q3_of_GNI - q1_of_GNI
print(IQR_GNI)


#outliers

south_asia_df['outlier'] = (
    (south_asia_df['hdi'] < (q1_of_HDI - 1.5 * IQR_HDI)) |
    (south_asia_df['hdi'] > (q3_of_HDI + 1.5 * IQR_HDI)) |
    (south_asia_df['gross_inc_percap'] < (q1_of_GNI - 1.5 * IQR_GNI)) |
    (south_asia_df['gross_inc_percap'] > (q3_of_GNI + 1.5 * IQR_GNI))
)

south_asia_df[['country', 'hdi', 'gross_inc_percap', 'outlier']]
```

Create a scatter plot of GNI per Capita vs HDI, highlighting the outliers in a different color.

```python
plt.figure()
plt.scatter(south_asia_df['gross_inc_percap'], south_asia_df['hdi'])

highlight_outliers = south_asia_df[south_asia_df['outlier']]
plt.scatter(highlight_outliers['gross_inc_percap'], highlight_outliers['hdi']
)

plt.xlabel("GNI per Capita")
plt.ylabel("HDI")
plt.title("HDI vs GNI per Capita of South Asia")
plt.show()
```

## HDI Disparity:

```python
#Compute the range (max – min) of HDI for each region.

SA_range = df_south_asia['hdi'].max() - df_south_asia['hdi'].min()
ME_range = df_middle_east['hdi'].max() - df_middle_east['hdi'].min()

print(f"Range of HDI for South Asia: {SA_range}")
print(f"Range of HDI for Middle East: {ME_range}")

#Compute the coefficient of variation (CV = std/mean) for HDI.

SA_coefV = df_south_asia['hdi'].std() / df_south_asia['hdi'].mean()
ME_coefV = df_middle_east['hdi'].std() / df_middle_east['hdi'].mean()

print(f"Coefficient of Variation (CV) for South Asia: {SA_coefV}")
print(f"Coefficient of Variation (CV) for Middle East: {ME_coefV}")

# 4. Identify which region exhibits more variation
if SA_coefV  > ME_coefV :
    print("\nSouth Asia exhibits more variation in HDI.")
else:
    print("\nThe Middle East exhibits more variation in HDI.")
```

## Outlier Detection:

```python
#Detect outliers in HDI {"hdi"} and GNI per Capita {"gross inc percap"} for each region using the 1.5 × IQR rule.

# South Asia
# HDI Outliers
Q1_sa_hdi = df_south_asia['hdi'].quantile(0.25)
Q3_sa_hdi = df_south_asia['hdi'].quantile(0.75)
IQR_sa_hdi = Q3_sa_hdi - Q1_sa_hdi
lower_sa_hdi = Q1_sa_hdi - 1.5 * IQR_sa_hdi
upper_sa_hdi = Q3_sa_hdi + 1.5 * IQR_sa_hdi

# GNI Outliers
Q1_sa_gni = df_south_asia['gross_inc_percap'].quantile(0.25)
Q3_sa_gni = df_south_asia['gross_inc_percap'].quantile(0.75)
IQR_sa_gni = Q3_sa_gni - Q1_sa_gni
lower_sa_gni = Q1_sa_gni - 1.5 * IQR_sa_gni
upper_sa_gni = Q3_sa_gni + 1.5 * IQR_sa_gni

# Middle East
# HDI Outliers
Q1_me_hdi = df_middle_east['hdi'].quantile(0.25)
Q3_me_hdi = df_middle_east['hdi'].quantile(0.75)
IQR_me_hdi = Q3_me_hdi - Q1_me_hdi
lower_me_hdi = Q1_me_hdi - 1.5 * IQR_me_hdi
upper_me_hdi = Q3_me_hdi + 1.5 * IQR_me_hdi

# GNI Outliers
Q1_me_gni = df_middle_east['gross_inc_percap'].quantile(0.25)
Q3_me_gni = df_middle_east['gross_inc_percap'].quantile(0.75)
IQR_me_gni = Q3_me_gni - Q1_me_gni
lower_me_gni = Q1_me_gni - 1.5 * IQR_me_gni
upper_me_gni = Q3_me_gni + 1.5 * IQR_me_gni
```

```python
# Scatter plot south asia for HDI vs GNI per Capita in the Middle East, highlighting outliers
#Creating the outlier dataframe for South Asia
sa_outliers = df_south_asia[
    (df_south_asia['hdi'] < lower_sa_hdi) | (df_south_asia['hdi'] > upper_sa_hdi) |
    (df_south_asia['gross_inc_percap'] < lower_sa_gni) | (df_south_asia['gross_inc_percap'] > upper_sa_gni)
]

plt.figure()
sns.scatterplot(data=df_south_asia, x='gross_inc_percap', y='hdi', color='blue', label='Normal Data')
sns.scatterplot(data=sa_outliers, x='gross_inc_percap', y='hdi', color='red', label='Outliers')
plt.xlabel("GNI per Capita")
plt.ylabel("HDI Score")
plt.title("South Asia: HDI vs GNI per Capita Outliers Highlighted")
plt.show()
```

```python
# Scatter plot middle east for HDI vs GNI per Capita in the Middle East, highlighting outliers

me_outliers = df_middle_east[
    (df_middle_east['hdi'] < lower_me_hdi) | (df_middle_east['hdi'] > upper_me_hdi) |
    (df_middle_east['gross_inc_percap'] < lower_me_gni) | (df_middle_east['gross_inc_percap'] > upper_me_gni)
]

plt.figure()
sns.scatterplot(data=df_middle_east, x='gross_inc_percap', y='hdi', color='green', label='Normal')
sns.scatterplot(data=me_outliers, x='gross_inc_percap', y='hdi', color='black', label='Outliers')

plt.title("Middle East: HDI vs GNI per Capita (Outliers Highlighted)")
plt.xlabel("GNI per Capita")
plt.ylabel("HDI Score")
plt.show()
```
